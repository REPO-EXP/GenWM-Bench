```python
import os


os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"


import sys
try:
    from datasets import load_dataset
    import requests
except ImportError:
    print("❌ pip install datasets requests")
    sys.exit()

def main():
    print("=" * 60)
    print("Stable-Diffusion-Prompts")
    print("=" * 60)


    dataset_name = "Gustavosta/Stable-Diffusion-Prompts"
    
    try:
        print(f"loading: {dataset_name}...")
        dataset = load_dataset(dataset_name, split="train", streaming=True)
        
        print("✅\n")
        
        count = 0
        for example in dataset:
            count += 1
            
            prompt = None
            potential_fields = ['Prompt', 'prompt', 'text', 'caption', 'description']
            
            for field in potential_fields:
                if field in example and isinstance(example[field], str):
                    prompt = example[field].strip()
                    break

            if not prompt:
                for val in example.values():
                    if isinstance(val, str) and len(val) > 5:
                        prompt = val.strip()
                        break

            print(f"No.{count:02d}")
            print("-" * 30)
            if prompt:
                display_text = prompt[:300] + "..." if len(prompt) > 300 else prompt
                print(display_text)
            else:
                print(f"original data: {str(example)[:100]}...")
            print()

            if count >= 40:
                break

        print("=" * 60)
        print(f"✅ totoal {count} data")

    except Exception as e:
        print(f"\n❌{e}")


if __name__ == "__main__":
    main()
```
