```python
import os
from pathlib import Path


TARGET_DIR = "."       
MAX_FILES_PER_DIR = 4  
MAX_DEPTH = 4         
EXCLUDED_NAMES = {
    "__init__.py", 
    "__pycache__", 
    ".git", 
    ".idea", 
    ".vscode", 
    ".DS_Store"
}


def generate_tree(directory, prefix="", current_depth=1):
    if current_depth > MAX_DEPTH:
        return

    path = Path(directory)
    
    try:
        items = sorted(list(path.iterdir()), key=lambda x: (x.is_file(), x.name.lower()))
    except PermissionError:
        print(f"{prefix}└── [Permission Denied]")
        return

    items = [item for item in items if item.name not in EXCLUDED_NAMES]

    dirs = [x for x in items if x.is_dir()]
    files = [x for x in items if x.is_file()]

    file_count = len(files)
    if file_count > MAX_FILES_PER_DIR:
        files_to_show = files[:MAX_FILES_PER_DIR]
        has_more_files = True
        omitted_count = file_count - MAX_FILES_PER_DIR
    else:
        files_to_show = files
        has_more_files = False
        omitted_count = 0

    display_items = dirs + files_to_show
    total_items = len(display_items)

    for index, item in enumerate(display_items):
        is_last_item = (index == total_items - 1) and not has_more_files
        connector = "└── " if is_last_item else "├── "
        
        print(f"{prefix}{connector}{item.name}")

        if item.is_dir():
            if current_depth < MAX_DEPTH:
                extension = "    " if is_last_item else "│   "
                generate_tree(item, prefix + extension, current_depth + 1)

    if has_more_files:
        print(f"{prefix}└── ... ({omitted_count} more files)")

if __name__ == "__main__":
    print(f"Structure for: {os.path.abspath(TARGET_DIR)}\n")
    print(os.path.basename(os.path.abspath(TARGET_DIR)) + "/")
    
    generate_tree(TARGET_DIR, current_depth=1)
```
